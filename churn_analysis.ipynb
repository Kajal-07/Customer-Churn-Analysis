```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("Customer_churn.csv")
df
```

```python
df.info()
```

```python
#replacing blanks with 0 as tenure is 0 and no total charges are recorded
```

```python
df['TotalCharges']=df['TotalCharges'].replace(" ","0")
df['TotalCharges']=df['TotalCharges'].astype("float")
```

```python
def conv(value):
    if value == 1:
        return "yes"
    else:
        return "no"

df['SeniorCitizen']=df['SeniorCitizen'].apply(conv)
```

```python
df.head(30)
```

```python
ax=sns.countplot(x='Churn',data=df)

ax.bar_label(ax.containers[0])
plt.title("Count of Customers by Churn")
plt.show()
```

```python
gb=df.groupby("Churn").agg({'Churn':'count'})
plt.pie(gb['Churn'],labels=gb.index,autopct="%%%1.2f")
plt.title("Percentage of Churned Custommers")
plt.show()
```

```python
sns.countplot(x="gender",data=df,hue='Churn')
plt.title("Churn by Gender")
plt.show()
```

```python
plt.figure(figsize=(3,3))
ax=sns.countplot(x="SeniorCitizen",data=df)
ax.bar_label(ax.containers[0])
plt.title("Count of Customers by Senior Citizens")
plt.xticks([0,1],['No','Yes'])
plt.show()
```

```python
# Calculate percentages within each SeniorCitizen category
churn_pct = pd.crosstab(
    df["SeniorCitizen"],
    df["Churn"],
    normalize="index"
) * 100

# Create stacked bar chart
ax = churn_pct.plot(
    kind="bar",
    stacked=True,
    figsize=(6, 4))
    

# Add percentage labels inside each segment
for container in ax.containers:
    labels = [
        f"{bar.get_height():.1f}%" if bar.get_height() > 0 else ""
        for bar in container
    ]
    ax.bar_label(container, labels=labels, label_type="center", fontsize=9)

plt.title("Churn Percentage by Senior Citizen Status")
plt.xlabel("Senior Citizen")
plt.ylabel("Percentage of Customers (%)")
plt.xticks([0, 1], ["No", "Yes"], rotation=0)
plt.ylim(0, 100)
plt.legend(title="Churn",bbox_to_anchor=(0.9,0.9))
plt.tight_layout()
plt.show()
```

```python
import textwrap
columns = [
    'PhoneService', 'MultipleLines', 'InternetService',
    'OnlineSecurity', 'OnlineBackup', 'DeviceProtection',
    'TechSupport', 'StreamingTV', 'StreamingMovies'
]

fig, axes = plt.subplots(3, 3, figsize=(20, 16))
axes = axes.flatten()

for i, col in enumerate(columns):
    sns.countplot(data=df, x=col, hue='Churn', ax=axes[i])

    axes[i].set_title(f'{col} vs Churn', fontsize=11)
    axes[i].set_xlabel('')
    axes[i].set_ylabel('Customer Count', fontsize=9)

    # Wrap long category labels and keep them horizontal
    labels = [
        textwrap.fill(label.get_text(), width=12)
        for label in axes[i].get_xticklabels()
    ]
    axes[i].set_xticks(
        axes[i].get_xticks(),
        labels,
        rotation=0,
        ha='center'
    )
    axes[i].tick_params(axis='x', labelsize=8, pad=5)

    # Remove individual legends
    legend = axes[i].get_legend()
    if legend is not None:
        legend.remove()

# Add one shared legend
handles, labels = axes[0].get_legend_handles_labels()
fig.legend(handles, labels, title='Churn',
           loc='center right', bbox_to_anchor=(0.99, 0.5))

fig.suptitle('Customer Services vs Churn Analysis', fontsize=16)

# Reserve space for the legend and title
fig.tight_layout(rect=[0, 0.02, 0.92, 0.95], h_pad=3, w_pad=2)

fig.savefig(
    'customer_services_churn.png',
    dpi=300,
    bbox_inches='tight'
)

plt.show()
```
